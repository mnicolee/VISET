# PR #66 review: two regressions

```python
import eunoia as eu

# made-up gene names, 7 characters each
genes = [f"GENE{i:03d}" for i in range(300)]
```

## 1. Hover text gets clipped

Hover over the Kinase inhibitors-only region: the gene list is wider than the 400px figure.

```python
sets = {
    "Statins": genes[0:90],
    "Kinase inhibitors": genes[60:300],
    "NSAIDs": genes[40:70] + genes[200:220],
}
fit = eu.euler(sets, seed=5)
fit.plot_plotly(width=400).show(config={"responsive": False})
```

```python
# hover font size per region (13 = default, smaller = shrunk to fit)
fig = fit.plot_plotly(width=400)
{t.name: t.hoverlabel.font.size for t in fig.data if t.hoverlabel.font.size}
```

## 1b. Too many names for a tiny figure

Should hit the min font size, show `+N more`, and print a warning.

```python
fit = eu.euler({"Statins": genes[0:90], "Kinase inhibitors": genes[60:300]}, seed=1)
fit.plot_plotly(width=200, height=150).show(config={"responsive": False})
```

No title, but there is an empty band at the top and the diagram is squeezed below it.

## 6. Title edge cases (possible problems)

### 6a. Long one-line title

Fixed: the default font shrinks until the title fits the width (15.5 → 10.5px).

```python
fit.plot_plotly(
    width=300, height=300, title="Drug targets shared by <br>statins and kinase inhibitors",
    quantities=False, hover=False,
).show(config={"responsive": False})
```

### 6b. Very short figure (height 30)

Fixed: used to crash (font < 1px). The default font now stops at 3px.

```python
fit.plot_plotly(width=300, height=30, title="Drug targets", quantities=False, hover=False).show(config={"responsive": False})
```

### 6c. Wide but short figure: title gets small

The 0.25 height cap shrinks the font to ~7px on a 1000px-wide figure. Expected.

```python
fit.plot_plotly(width=1000, height=100, title="Drug targets", quantities=False, hover=False).show(config={"responsive": False})
```

### 6d. Huge user font eats the whole figure

User `fontsize` is used as-is, so the diagram shrinks to ~1px. User's choice.

```python
fit.plot_plotly(
    width=300, height=150, title={"text": "Drug targets", "fontsize": 200},
    quantities=False, hover=False,
).show(config={"responsive": False})
```

### 6e. `<BR>` (uppercase) line break

Fixed: line breaks are now counted the way plotly does (`<br>`, `<BR>`, `<br >`), so the band fits both lines.

```python
fit.plot_plotly(
    width=300, height=150, title="Drug targets<br>statins vs kinase inhibitors",
    quantities=False, hover=False,
).show(config={"responsive": False})
```
