# eunoia with a sample long-name dataset
Same tests as `eunoia_contribution.ipynb` A2, but with the sample long-name dataset in `samples/` (made-up names): gene-style names up to 12 chars and protein-style names up to 58 chars. Regenerate it with `python samples/make_sample_long_names.py`.

```python
import warnings
from pathlib import Path

import matplotlib.pyplot as plt
import pandas as pd
import eunoia as eu

print("eunoia", eu.__version__)

# show every "did not fit" warning, not just the first one per cell
warnings.simplefilter("always", UserWarning)

DATA = Path("samples")  # sample long-name dataset (made-up names)
FILES = {
    "Drug A": "sample_long_names_DrugA.csv",
    "Drug B": "sample_long_names_DrugB.csv",
    "Drug C": "sample_long_names_DrugC.csv",
}
DF = {k: pd.read_csv(DATA / f) for k, f in FILES.items()}

# gene names (short-ish) and protein names (long; first name only if several are listed)
GENES = {k: df["Gene names"].dropna().astype(str).tolist() for k, df in DF.items()}
PROTEINS = {k: df["Protein names"].dropna().astype(str).str.split(";").str[0].tolist() for k, df in DF.items()}

for k in FILES:
    print(f"{k:9} {len(GENES[k]):3} genes (max {max(map(len, GENES[k]))} chars), "
          f"proteins max {max(map(len, PROTEINS[k]))} chars")
```

## B1. Gene names, packed (default)

```python
fit = eu.euler({"Drug A": GENES["Drug A"], "Drug B": GENES["Drug B"], "Drug C": GENES["Drug C"]}, seed=1)
fig, ax = plt.subplots(figsize=(10, 10))
fit.plot(ax=ax, members={"mode": "packed", "fontsize": 28}, labels={"fontsize": 38})
ax.set_title("Gene names overlap_B")
plt.savefig("images/Gene names overlap_B.png", dpi=200, bbox_inches="tight")
plt.show()
```

## O. Overflow: which names go missing? Default font vs IBM Plex Mono vs Inter

```python
# register the fonts in VISET/Font so matplotlib can use them by name
from matplotlib import font_manager

FONTS = Path("../../Font")
font_manager.fontManager.addfont(str(FONTS / "IBM_Plex_Mono" / "IBMPlexMono-Regular.ttf"))
font_manager.fontManager.addfont(str(FONTS / "Inter" / "Inter-Regular.ttf"))
# default font is "DejaVu Sans"; now "IBM Plex Mono" and "Inter" work too
```

```python
fit = eu.euler({"Drug A": GENES["Drug A"], "Drug B": GENES["Drug B"], "Drug C": GENES["Drug C"]}, seed=1)
fig, ax = plt.subplots(figsize=(10, 10))
with warnings.catch_warnings(record=True) as caught:
    warnings.simplefilter("always")   # fresh warnings every run, not once per session
    fit.plot(ax=ax, members={"mode": "packed", "fontfamily": "IBM Plex Mono", "fontsize": 28}, labels={"fontsize": 38})
ax.set_title("O1 Sample genes 10x10, IBM Plex Mono")
#plt.savefig("images/Sample O1 genes 10x10 DejaVu Sans.png", dpi=200, bbox_inches="tight")
plt.show()

# which names are missing, from eunoia's own "did not fit" count
# (eunoia stops at the first name that doesn't fit, so the missing ones are the last N in each region's list)
import ast
dropped = {}
for w in caught:
    msg = str(w.message)
    print("WARNING:", msg)
    if "did not fit" in msg:
        dropped = ast.literal_eval(msg.split(": ", 1)[1])
for region, n in dropped.items():
    print(region, "missing:", fit.members[region][-n:])

# regions with members that eunoia didn't draw at all (zero area, never in the warning)
for region, names in fit.members.items():
    if names and fit.fitted_values.get(region, 0) == 0:
        print(region, "not drawn:", names)

print("fonts drawn:", {t.get_fontname() for t in ax.texts})
```

## P. Interactive (plotly)
`width` / `height` are in pixels (height defaults to keep the diagram's shape). Hover a region to see its members. Zoom with the toolbar: text keeps its pixel size, so zooming in spreads names apart.

### P1. Gene names, default width (700 px)

```python
fit = eu.euler({"Drug A": GENES["Drug A"], "Drug B": GENES["Drug B"], "Drug C": GENES["Drug C"]}, seed=1)
fig = fit.plot_plotly(width=700, height=700)
fig.update_layout(title="P1 Sample genes, width 700", margin={"t":50})
fig.show()
```

### P2. Gene names, wide (1400 px)

```python
fit = eu.euler({"Drug A": GENES["Drug A"], "Drug B": GENES["Drug B"], "Drug C": GENES["Drug C"]}, seed=1)
fig = fit.plot_plotly(width=1400, height=None)
fig.update_layout(title="P2 Sample genes, width 1400")
fig.show()
```
